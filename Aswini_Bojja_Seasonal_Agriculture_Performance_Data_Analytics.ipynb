# 9.Outlier Analysis
Investigate potential outliers using statistical and visual methods.

**Univariate: Season distribution(Pie Chart)**

```python
plt.figure(figsize=(7, 7))
df['Season'].value_counts().plot.pie(autopct='%1.1f%%', startangle=90,cmap='viridis')
plt.title("Distribution of Records by Season")
plt.ylabel("") #Hide the default 'Season' label on the y-axis
plt.tight_layout()
plt.show()
```

```python
numeric_columns = df.select_dtypes(include=np.number).columns.tolist()

# IQR -based outlier summary for numerical columns
outlier_summary = []

for col in numeric_columns:
  q1 = df[col].quantile(0.25)
  q3 = df[col].quantile(0.75)
  iqr = q3 - q1
  lower = q1 - 1.5 * iqr
  upper = q3 + 1.5 * iqr
  count = ((df[col] < lower) | (df[col] > upper)).sum()

  outlier_summary.append({
      "Column": col,
      "Lower_Bound": lower,
      "Upper_Bound": upper,
      "Outlier_Count": count,
      "Outlier_Percentage": round(count / len(df) * 100, 2)
  })

outlier_summary = pd.DataFrame(outlier_summary).sort_values(
    "Outlier_Count", ascending=False
)
outlier_summary
```

## 16. Conclusion

The analysis of the **Seasonal Agriculture Performance Dataset** revealed several important patterns in agricultural yield and profitability.

The analysis showed that **Kharif season had the highest average yield (5.64 tonnes/ha)** and the highest average profit. Among irrigation methods, **Drip irrigation had the highest average yield (6.62 tonnes/ha) and average profit**. The analysis also found a **strong positive association between water-use efficiency and crop yield**, while seed quality score showed almost no linear relationship with yield.

Another important finding was that **Punjab had the highest average yield among the states**, and **Sugarcane had the highest average yield among the crops** in the dataset.

However, the analysis has some limitations. The results mainly show **associations and patterns, not cause-and-effect relationships**. Yield and profit can be influenced by several factors such as crop type, rainfall, soil conditions, irrigation, costs, and market prices. Therefore, the findings should not be interpreted as proof that a particular factor directly causes higher performance.

Overall, this analysis provides useful evidence for understanding **seasonal agricultural performance, irrigation practices, water efficiency, crop performance, and profitability**. These findings can support future investigation by examining the relationships between multiple agricultural factors and can help guide data-driven agricultural planning and decision-making.
